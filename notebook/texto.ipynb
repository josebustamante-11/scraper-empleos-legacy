```python
import re

def obtener_sueldo(texto: str) -> float:
    """
    Extrae el primer monto en soles (S/) del texto y lo convierte en float.
    Devuelve None si no encuentra ningún monto.
    """
    patron = r"S/\s*([\d.,]+)"
    match = re.search(patron, texto)

    if not match:
        return None

    monto = match.group(1)

    # Eliminar separadores de miles y normalizar
    monto = monto.replace(",", "").replace(".", "")

    return float(monto)
```
